```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# verif categorial data look clean

cat_cols = ['Industry','Founder_Type','Funding_Stage',
            'Economic_Climate','Burnout_Level','Shutdown_Risk','Seeks_Mental_Health_Support',"Work_Mode"]
for col in cat_cols:
    print(col, df[col].unique())
```

```python
# to check outliers

cols = ['Weekly_Work_Hours','Stress_Score','Burnout_Score',
        'Monthly_Revenue_Growth_Percent','Shutdown_Probability']
df[cols].hist(bins=30, figsize=(14,8))
plt.tight_layout()
plt.show()
```

```python
failed = df[df['Startup_Failure_Flag']==1]
survived = df[df['Startup_Failure_Flag']==0]

compare = pd.DataFrame({
  'Failed': failed[['Stress_Score','Burnout_Score',
    'Weekly_Work_Hours','Runway_Months_Remaining']].mean(),
  'Survived': survived[['Stress_Score','Burnout_Score',
    'Weekly_Work_Hours','Runway_Months_Remaining']].mean()
})
print(compare)
```

```python
#economic climate vs failurer rate
climate = df.groupby('Economic_Climate').agg(
  Failure_Rate=('Startup_Failure_Flag','mean'),
  Avg_Revenue_Growth=('Monthly_Revenue_Growth_Percent','mean'),
  Avg_Runway=('Runway_Months_Remaining','mean')
).round(2)
print(climate.sort_values('Failure_Rate', ascending=False))
```

```python
#funding stage vs survival
funding = df.groupby('Funding_Stage').agg(
  Failure_Rate=('Startup_Failure_Flag','mean'),
  Avg_Runway=('Runway_Months_Remaining','mean')
).round(3)
print(funding.sort_values('Failure_Rate', ascending=False))
```

```python
#Work hours vs revenue growth
df['Hours_Bucket'] = pd.cut(df['Weekly_Work_Hours'],
  bins=[0,40,55,70,85,120],
  labels=['<40','40-55','55-70','70-85','85+'])

hours_rev = df.groupby('Hours_Bucket')['Monthly_Revenue_Growth_Percent'].mean()
sns.barplot(x=hours_rev.index, y=hours_rev.values)
plt.title('Avg Revenue Growth by Weekly Work Hours')
plt.show()
```
